# Chunking Strategies — Part 2

A **chunk** is the unit of retrieval: when someone asks a question, search returns *chunks*,
and the agent only sees what's inside those chunks. How we cut the document decides, before any
model gets involved, **what is possible to answer**.

**The tension:**
- chunk **too big** → the embedding becomes an average of several topics, weakly similar to any
  one specific question; and the agent gets irrelevant junk dumped alongside the real answer.
- chunk **too small** → the answer ends up spread across several chunks and search only returns
  one of them; a chunk on its own loses the context that made it meaningful
  (`"- 12 months"` of *what*?).
- target: **one chunk ≈ one self-contained idea**.

**Three ways to decide where to cut**, compared here on the same two documents:
- **fixed** — counts characters, cuts every N (with M characters of overlap), never looks at
  the content.
- **markdown** — cuts on the headings (`#`, `##`, `###`) the author already wrote.
- **semantic** — embeds sentence by sentence and cuts where the meaning shifts the most.


```python
import sys, time
from pathlib import Path

ROOT = Path.cwd().resolve().parents[1]  # docs/videos/ -> repo root
sys.path.insert(0, str(ROOT))

import tiktoken
from rag.loaders import split_fixed, split_markdown, split_semantic
from rag.ingest import get_embedder

enc = tiktoken.get_encoding("cl100k_base")

def n_tokens(text: str) -> int:
    return len(enc.encode(text))
```

```python
produto_a_path = ROOT / "kb/empresa_x/products/produto_a.md"
lead_maria_path = ROOT / "kb/empresa_x/lead_reports/lead_maria.md"

produto_a = produto_a_path.read_text()
lead_maria = lead_maria_path.read_text()

for name, text in [("produto_a.md", produto_a), ("lead_maria.md", lead_maria)]:
    print(f"=== {name} — {len(text)} chars / {n_tokens(text)} tokens ===")
    print(text)
    print()
```

```python
def show_chunks(chunks: list[str]) -> None:
    for i, c in enumerate(chunks):
        print(f"--- chunk {i} — {len(c)} chars / {n_tokens(c)} tokens ---")
        print(c)
        print()
    print(f"=== total: {len(chunks)} chunks ===")
```

## Fixed

Counts characters and cuts every `chunk_size` characters (with `chunk_overlap` characters
repeated between neighboring chunks), never looking at the text's content.


```python
show_chunks(split_fixed(produto_a, chunk_size=800, chunk_overlap=200))
```

```python
show_chunks(split_fixed(lead_maria, chunk_size=800, chunk_overlap=200))
```

**Notice:**
- `produto_a.md` is 779 chars — smaller than `chunk_size=800` — so it becomes **1 chunk = the
  entire file**. A `chunk_size` bigger than the document makes "fixed" do nothing, silently.
- `lead_maria.md` (longer) already splits into several chunks, and at least one cut lands **in
  the middle of a section** (look at where "Context and Objectives" gets cut) — "fixed" has no
  idea a topic boundary exists there.


```python
# Change these two numbers live and compare this output with the one above.
show_chunks(split_fixed(lead_maria, chunk_size=300, chunk_overlap=50))
```

**Notice:** with a smaller `chunk_size`, some chunks are now just a bare heading (e.g.
`"## Context and Objectives"` alone, without the text that follows) and sentences get cut mid-way.
The text repeated between the end of one chunk and the start of the next is `chunk_overlap` in
action — it exists to give a second chance to any sentence that happens to land right on the cut.


## Markdown

Cuts on the headings the author already wrote (`#`, `##`, `###`) — assumes the document's
structure already is the structure of its meaning.


```python
show_chunks(split_markdown(produto_a))
```

```python
show_chunks(split_markdown(lead_maria))
```

**Notice:** each chunk starts with its `h1 > h2 > h3` breadcrumb (e.g.
`"Product A — Specifications and Policies (Empresa X) > Warranty and Support"`) before the body
text. `MarkdownHeaderTextSplitter` moves the heading itself into `Document.metadata` and leaves
only the body in `page_content` — without reconstructing it back into the text, a chunk like
*Warranty and Support*'s body would never contain the word "Warranty" at all, and the first
chunk of `lead_maria` would never say "Maria". `split_markdown`/`split_markdown_hybrid`
(`rag/loaders.py`) prepend this breadcrumb explicitly so every chunk stays self-contained — see
`docs/markdown_header_drop.md` for the full writeup of why this matters and what it took to fix.


## Semantic

Embeds sentence by sentence and cuts where two consecutive sentences are most dissimilar —
adapts to the content, but costs one API call per document and isn't deterministic (the cut can
move a little between runs).


```python
embedder = get_embedder()

t0 = time.time()
chunks_produto_a_sem = split_semantic(produto_a, embedder)
t1 = time.time()
print(f"produto_a.md — {t1 - t0:.2f}s")
show_chunks(chunks_produto_a_sem)
```

```python
t0 = time.time()
chunks_lead_maria_sem = split_semantic(lead_maria, embedder)
t1 = time.time()
print(f"lead_maria.md — {t1 - t0:.2f}s")
show_chunks(chunks_lead_maria_sem)
```

**Notice:** where the cut landed *inside* a section, at a point where the topic shifts but
there's no heading there — this is the one strategy of the three capable of that. Compare this
cell's runtime with the instant execution of `fixed`/`markdown`: that's the cost difference.


```python
def stats(chunks: list[str]) -> dict:
    sizes = [len(c) for c in chunks]
    toks = [n_tokens(c) for c in chunks]
    return {
        "n_chunks": len(chunks),
        "chars_min": min(sizes), "chars_mean": round(sum(sizes) / len(sizes)), "chars_max": max(sizes),
        "tokens_mean": round(sum(toks) / len(toks)),
    }

rows = {
    ("produto_a", "fixed"): stats(split_fixed(produto_a, 800, 200)),
    ("produto_a", "markdown"): stats(split_markdown(produto_a)),
    ("produto_a", "semantic"): stats(chunks_produto_a_sem),
    ("lead_maria", "fixed"): stats(split_fixed(lead_maria, 800, 200)),
    ("lead_maria", "markdown"): stats(split_markdown(lead_maria)),
    ("lead_maria", "semantic"): stats(chunks_lead_maria_sem),
}

header = f"{'doc':<12}{'strategy':<10}{'n_chunks':>10}{'chars_min':>11}{'chars_mean':>12}{'chars_max':>11}{'tokens_mean':>13}"
print(header)
print("-" * len(header))
for (doc, strat), s in rows.items():
    print(f"{doc:<12}{strat:<10}{s['n_chunks']:>10}{s['chars_min']:>11}{s['chars_mean']:>12}{s['chars_max']:>11}{s['tokens_mean']:>13}")
```

## The bridge to the agent

Real question: **"What are the commercial terms and warranty for Product A?"**
A correct answer needs **both** facts: `"30/60"` (installment plan) and `"12 months"`
(warranty). This is exactly what the agent receives as context — nothing beyond the chunks
search chose to hand it.

Marking by hand, per strategy, which chunk(s) of `produto_a.md` each fact lands in:

- **fixed (800/200):** 1 chunk = the entire file → both facts sit together, but come bundled
  with 3 irrelevant sections (Overview, Included Items, Notes and Upgrades) the agent also
  receives for no reason.
- **markdown:** `"30/60"` is in the *Commercial Terms* chunk, `"12 months"` is in the *Warranty
  and Support* chunk — **two different chunks**. If search only returns the single best chunk
  (top-1), the agent gets only half the answer. The breadcrumb fix means it at least knows which
  half — the surviving chunk still says "Warranty and Support" — but the fact that the two facts
  live in different chunks at all is still down to how this document's author split them across
  headings, not something the fix changes.
- **semantic:** depends on where the cut landed in this run — check above whether the two facts
  ended up together or apart.

**The point:** the chunking strategy already decided whether this question is answerable, before
search even runs. This is what the agent "sees" — not the whole document, only the chunks the
chosen strategy produced.
