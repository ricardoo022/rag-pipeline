# Search Strategies — Part 3

Chunking (Part 2) decided what is *possible* to retrieve. **Search** decides what actually
comes back for a given question.

Two fundamentally different ways to search the same KB:
- **vector** — match by *meaning*: embed the question, find the chunk vectors with the smallest
  cosine distance.
- **text** — match by *exact words*: the question becomes a set of stemmed words ANDed together,
  matched against a precomputed full-text column.

Each alone has a blind spot the other covers. So you run both and merge the two result lists —
that's **hybrid** search, and there are two ways to merge:
- **hybrid_rrf** — merge by *rank position* (Reciprocal Rank Fusion). The one to ship.
- **hybrid_union** — merge by *raw score*. The naive version, kept here only for contrast.

Two questions are carried through the whole notebook:
- `Q_MEANING` — phrased differently from the document, non-numeric answer. Tests
  meaning-matching in isolation (a numeric answer would also test "are numbers a weak vector
  signal?", which is a *different* question — kept out of this pair on purpose).
- `Q_EXACT` — the answer is an exact token/short phrase that appears verbatim in the document.
  Tests word-matching.

**Prerequisite:** semantic chunks for Empresa X must be materialized first:
`make rag_phase0 STRATEGY=semantic EMPRESA=empresa_x`

```python
import sys
from pathlib import Path

ROOT = Path.cwd().resolve().parents[1]  # docs/videos/ -> repo root
sys.path.insert(0, str(ROOT))

from rag.search import kb_search_client  # .env is loaded automatically via rag.db

EMPRESA = "Empresa X"
CHUNKING = "semantic"

Q_MEANING = "What kind of assistance does Service B provide when something goes wrong?"  # answer: "N2 support"
Q_EXACT   = "What is the priority support channel for Service B?"                          # answer: "portal"
```

```python
def show_results(results: list[dict]) -> None:
    """Print a ranked result list: position, doc#chunk, score, snippet."""
    if not results:
        print("(no results)")
        return
    for rank, r in enumerate(results, start=1):
        doc = Path(r["doc_path"]).name if r.get("doc_path") else "-"
        score = r.get("score")
        score_s = f"{score:.4f}" if isinstance(score, (int, float)) else "-"
        snippet = " ".join((r.get("content") or "").split())[:160]
        print(f"#{rank}  {doc}#{r.get('chunk_ix')}  score={score_s}")
        print(f"     {snippet}")
    print(f"--- {len(results)} result(s) ---")


def search(query: str, search_type: str, **kw):
    return kb_search_client.invoke({
        "query": query,
        "k": 5,
        "search_type": search_type,
        "reranker": "none",
        "empresa": EMPRESA,
        "chunking": CHUNKING,
        **kw,
    })
```

## 1. Vector search — match by meaning

Embed the question with the same model used at ingestion, then sort every chunk by cosine
distance to that query vector (`order by embedding <=> qvec asc` in `sql/03_functions.sql:33`).
Closest = most similar meaning. The reported `score` is `1 - distance`, so ~1.0 means
near-identical meaning. `match_threshold=0.50` drops anything below 0.50 similarity.

```python
print("Q_MEANING:", Q_MEANING)
show_results(search(Q_MEANING, "vector", match_threshold=0.50))
```

```python
print("Q_EXACT:", Q_EXACT)
show_results(search(Q_EXACT, "vector", match_threshold=0.50))
```

**Notice:** vector search should still return the "N2 support" chunk for `Q_MEANING` even though
almost no words are shared with the document — "assistance" vs. "support", "when something goes
wrong" vs. "incidents and requests." That's the whole point: vector matches the *topic*, not the
wording. `Q_EXACT`'s answer ("portal") is a short, unambiguous word, so vector should find it
easily too here.

Neither question hinges on an exact number on purpose — that keeps this pair a clean test of
"does vector survive a vocabulary mismatch?" without also dragging in "vector is weak on exact
numbers," which is a separate, real limitation. If you want to see *that* one directly, ask a
question whose answer is a number (e.g. the `99.9%` SLA) and watch how close its score is to a
neighboring, wrong sentence.

## 2. Text search — match by exact words

`plainto_tsquery('english', question)` turns the question into stemmed words joined by AND
(`sql/03_functions.sql:58`). `fts @@ tsq` (line 64) keeps only chunks that contain **every** one
of those words. `ts_rank_cd` (line 65) ranks the survivors. No embedding, no API call, instant —
but no fuzziness at all.

```python
print("Q_MEANING:", Q_MEANING)
show_results(search(Q_MEANING, "text"))
```

```python
print("Q_EXACT:", Q_EXACT)
show_results(search(Q_EXACT, "text"))
```

```python
# What does the question actually become?
from rag.db import get_conn

with get_conn() as conn, conn.cursor() as cur:
    for q in (Q_MEANING, Q_EXACT):
        cur.execute("select plainto_tsquery('english', %s)::text", (q,))
        print(f"{q!r}\n  -> {cur.fetchone()[0]}\n")
```

**Notice:** `Q_MEANING` ANDs together stems like `assist`, `wrong`, `someth` — words that never
appear in the document (it says "support" and "incidents", not "assistance" and "wrong"). Since
every term must match, the result is an **empty list** — not bad results, *nothing*. `Q_EXACT`'s
words ("priority", "support", "channel", "service") are all in the document, so text search is
sharp and correct. Text is a specialist for exact tokens and useless for paraphrase — which is
why it is never run alone, only as the keyword half of hybrid.

## 3. Hybrid — RRF (merge by rank)

Run both searches, then merge by **position**, not score: `1.5/(60+vec_rank) +
1.0/(60+text_rank)` (`sql/03_functions.sql:105`). A chunk that appears in *both* lists gets both
contributions added together, so agreement between the two methods floats to the top.

```python
print("Q_MEANING:", Q_MEANING)
show_results(search(Q_MEANING, "hybrid_rrf", match_threshold=0.50))
```

```python
print("Q_EXACT:", Q_EXACT)
show_results(search(Q_EXACT, "hybrid_rrf", match_threshold=0.50))
```

**Notice:** both questions now return the answer chunk near the top. The vector half carries
`Q_MEANING` (text found nothing), the text half sharpens `Q_EXACT`. Neither question depends on a
single method anymore.

## 4. Hybrid — union (merge by raw score)

The naive merge: concatenate both result lists and `order by score desc`
(`sql/03_functions.sql:142`). But vector scores are ~0.7–0.9 and text scores are ~0.02–0.1, so
the vector rows always sit on top regardless of how good the text hits were.

```python
print("Q_MEANING:", Q_MEANING)
show_results(search(Q_MEANING, "hybrid_union", match_threshold=0.50))
```

```python
print("Q_EXACT:", Q_EXACT)
show_results(search(Q_EXACT, "hybrid_union", match_threshold=0.50))
```

**Notice:** compare this ordering to the RRF section. Union sorts by a `score` column that means
different things for the two halves, so the text contribution barely moves anything — you have
essentially run vector search with a few text leftovers appended.

## RRF vs union, side by side

One query, both input rankings and both merged outputs, as `(doc, chunk_ix)` tuples so the
ordering is easy to compare.

```python
q = Q_EXACT

vec = search(q, "vector", match_threshold=0.50)
txt = search(q, "text")
rrf = search(q, "hybrid_rrf", match_threshold=0.50)
uni = search(q, "hybrid_union", match_threshold=0.50)

def keys(results):
    return [(Path(r["doc_path"]).name, r["chunk_ix"]) for r in results]

print("QUERY:", q, "\n")
print("VECTOR ranking :", keys(vec))
print("TEXT   ranking :", keys(txt))
print()
print("UNION  merged  :", keys(uni))
print("RRF    merged  :", keys(rrf))
```

## 5. Add-ons — HyDE and rerank

Both are orthogonal to the four search types above: pick one of `vector` / `text` /
`hybrid_rrf` / `hybrid_union`, then optionally bolt either of these on top. Neither one is a
search method by itself.

### HyDE — fixes the question, before the vector search runs

`use_hyde=True` (`rag/search.py:187–190`): instead of embedding the raw question, an LLM first
writes a short hypothetical answer paragraph, and *that* gets embedded. It only touches the
vector side of a search — it's silently ignored when `search_type="text"`.

```python
from rag.search import hyde

hyde_doc = hyde(Q_MEANING)
print("HyDE hypothetical answer for Q_MEANING:\n")
print(hyde_doc)
```

**Notice:** the LLM never saw our documents — it just imagined what a plausible answer would
sound like. If it's any good, that imagined paragraph uses vocabulary closer to the real
document ("support", "incidents", "service") than the original question did ("assistance",
"something goes wrong"). That's the whole trick: embed the imagined answer instead of the
question, and the vocabulary gap shrinks.

```python
print("VECTOR, Q_MEANING, without HyDE:")
show_results(search(Q_MEANING, "vector", match_threshold=0.50))

print("\nVECTOR, Q_MEANING, with HyDE:")
show_results(search(Q_MEANING, "vector", match_threshold=0.50, use_hyde=True))
```

**Notice:** compare the ranking and the scores between the two runs above — did the right chunk
move up, and by how much? Two costs to say out loud: one extra LLM call *before* every search,
and HyDE's output isn't deterministic, so the ranking can shift slightly between runs.

```python
# Proof that HyDE is a no-op for text search — these two calls should return identical results.
print("TEXT, Q_MEANING, use_hyde=False:")
show_results(search(Q_MEANING, "text"))

print("\nTEXT, Q_MEANING, use_hyde=True (ignored — no embedding is computed for text search):")
show_results(search(Q_MEANING, "text", use_hyde=True))
```

### Rerank — fixes the order, after retrieval

`reranker="cohere"` (`apply_rerank`, `rag/search.py:86`): pull a wider net of candidates
(`rerank_candidates`, default 24) instead of just `k`, then a Cohere cross-encoder scores each
`(query, chunk)` pair *together* and the top `k` of that new ordering is returned. Needs
`COHERE_API_KEY` + `langchain-cohere`; a trial key can hit a 429, so this cell is wrapped in a
`try/except` rather than assumed to succeed.

```python
print("HYBRID_RRF, Q_EXACT, without rerank:")
show_results(search(Q_EXACT, "hybrid_rrf", match_threshold=0.50))

print("\nHYBRID_RRF, Q_EXACT, with Cohere rerank:")
try:
    show_results(search(Q_EXACT, "hybrid_rrf", match_threshold=0.50, reranker="cohere"))
except Exception as e:
    print(f"(rerank unavailable: {e})")
```

**Notice:** rerank can only **reorder** the 24 candidates `hybrid_rrf` already retrieved — it
can't pull in a chunk that retrieval missed entirely. So it's a precision tool for the top of the
list (`hit@1`), not a recall tool. Compare the top-1 result with and without it; if they're
already the same chunk, rerank had nothing to fix here and you'd need a harder query to see it
move something.

## The bridge to the eval

Two questions, by eye, is not a result. `tests/rag/experiments.yaml` sweeps these `search_type`
values over the **full** `rag_queries.yaml` set and `eval/run.py` scores each with
`hit@1 / hit@5 / judge_yes` — the same harness as Part 2, just changing `search_type` instead of
`chunking`. Note that `text` is never swept alone there: it only exists as the keyword half
inside `hybrid_rrf` / `hybrid_union`. `use_hyde` and `reranker` are columns in the same
`experiments.yaml` schema, so HyDE/rerank rows can be added to the same sweep rather than judged
by eye on two questions.

Run it: `PYTHONPATH=. .venv/bin/python eval/run.py --fast`
